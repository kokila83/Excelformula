```python

import gradio as gr
from google.colab import userdata
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

# 1. Retrieve credentials from Colab Secrets
api_key = userdata.get('AZURE_AI_FOUNDRY_KEY')
endpoint = "https://ganapathy-0333-resource.services.ai.azure.com/models"

# 2. Initialize Azure AI Inference Client
client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(api_key)
)

# 3. Define response function
def generate_response(prompt, system_instruction):
    if not prompt.strip():
        return "Please enter a message."

    try:
        response = client.complete(
            messages=[
                SystemMessage(content=system_instruction),
                UserMessage(content=prompt)
            ],
            model="gpt-4o-mini",  # Your deployment name from Azure Foundry
            temperature=0.7
        )
        return response.choices[0].message.content
    except Exception as e:
        return f"Error: {str(e)}"

# 4. Create Gradio Application UI
with gr.Blocks() as app:
    gr.Markdown("# 🤖 Azure AI Foundry Assistant")

    system_input = gr.Textbox(
        label="System Instruction",
        value="You are a helpful AI assistant.",
        lines=2
    )
    user_input = gr.Textbox(
        label="Your Message",
        placeholder="Type your prompt here...",
        lines=3
    )

    submit_btn = gr.Button("Send", variant="primary")
    output_text = gr.Textbox(label="AI Response", lines=6)

    submit_btn.click(
        fn=generate_response,
        inputs=[user_input, system_input],
        outputs=output_text
    )

# 5. Launch web application
app.launch(theme=gr.themes.Soft(), share=True, debug=True)
```
