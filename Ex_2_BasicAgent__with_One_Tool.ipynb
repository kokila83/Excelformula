```python
import getpass
from openai import OpenAI

# 1. Enter the API Key copied directly from the Azure AI Foundry portal
api_key = getpass.getpass("Enter your Azure AI Foundry API key: ")

# 2. Configure client with your project's OpenAI-compatible base_url
client = OpenAI(
    api_key=api_key.strip(),  # .strip() removes accidental surrounding whitespace
    base_url="https://ganapathy-0333-resource.services.ai.azure.com/openai/v1/"
)

# 3. Define the tool
def calculator(expression):
    try:
        result = eval(expression)
        return f"Result: {result}"
    except Exception as e:
        return f"Error in calculation: {e}"

# 4. Agent query
user_query = "What is 15% of 2500?"

try:
    # 5. Send completion request using your exact deployment name
    analysis = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[
            {
                "role": "user",
                "content": f"""User asked: {user_query}
Available tool: calculator(expression)

Do you need the calculator? Respond with EXACTLY one of these formats:
USE_TOOL: <python math expression>
NO_TOOL: <direct answer>"""
            }
        ]
    )

    response_text = analysis.choices[0].message.content.strip()
    print(f"\nAgent decision:\n{response_text}\n")

    # 6. Parse decision and run tool if requested
    if "USE_TOOL:" in response_text:
        expression = response_text.split("USE_TOOL:")[1].strip()
        result = calculator(expression)
        print(f"Agent executed tool: calculator({expression}) -> {result}")

except Exception as e:
    print(f"\nAuthentication or Execution Error: {e}")
```
